*Exercise 8:* Write a function that is called with a board and player number, takes input from the player using Python's `input()`, and modifies the board using your function from Exercise 7.

Keep asking for input until the player enters a valid location that results in a valid move.


````python
def get_player_move(board, player):
    n = len(board)
    m = len(board[0])
    while True:
        location = input(f"Player {player}, enter a location (e.g. A1): ").strip()
        if len(location) < 2 or not location[0].isalpha() or not location[1:].isdigit():
            print("Invalid format. Please enter a letter followed by a number.")
            continue
        row = ord(location[0].upper()) - ord('A')
        col = int(location[1:]) - 1
        if not (0 <= row < n) or not (0 <= col < m):
            print(f"Out of range. Rows must be A-{chr(ord('A') + n - 1)}, columns 1-{m}.")
            continue
        success = place_mark_at_location(board, player, location)
        if success:
            print(f"Player {player} placed a mark at {location.upper()}.")
            return
        else:
            print("That cell is already taken. Try again.")
````

````python
def computer_move(board, computer_player, human_player):
    n = len(board)
    for row in range(n):
        for col in range(n):
            if board[row][col] == 0:
                board[row][col] = computer_player       
                if winner(board) == computer_player:
                    return                              
                board[row][col] = 0                    

    for row in range(n):
        for col in range(n):
            if board[row][col] == 0:
                board[row][col] = human_player           
                human_would_win = winner(board) == human_player
                board[row][col] = 0                   
                if human_would_win:
                    board[row][col] = computer_player     
                    return

    for row in range(n):
        for col in range(n):
            if board[row][col] == 0:
                board[row][col] = computer_player
                return

def play_game_vs_computer(n=3, human_player=1, computer_player=2):
    board = game_board(n)
    current_player = 1
    while True:
        draw_game_board(board)
        if current_player == human_player:
            get_player_move(board, current_player)
        else:
            computer_move(board, computer_player, human_player)

        status = winner(board)
        if status == human_player:
            draw_game_board(board)
            print("You win!")
            break
        elif status == computer_player:
            draw_game_board(board)
            print("Computer wins!")
            break
        elif status == 0:
            draw_game_board(board)
            print("It's a draw!")
            break
        else:
            current_player = 2 if current_player == 1 else 1
````

*Exercise 13:* Make the 2 computer players play each-other for 10 games on a 3x3, then 4x4, then 5x5 grid. Set the max depth so that the games only take seconds. Measure the "smarter" player's win rate for each grid.

````python
def evaluate_line(line, player, opponent):
    if opponent not in line:
        count = line.count(player)
        if count > 0:
            return count ** 2
    return 0

def heuristic_score(board, player, opponent):
    n = len(board)
    score = 0
    for row in board:
        score += evaluate_line(row, player, opponent) - evaluate_line(row, opponent, player)
    for col in range(n):
        column = [board[r][col] for r in range(n)]
        score += evaluate_line(column, player, opponent) - evaluate_line(column, opponent, player)
    diag = [board[i][i] for i in range(n)]
    score += evaluate_line(diag, player, opponent) - evaluate_line(diag, opponent, player)
    anti = [board[i][n - 1 - i] for i in range(n)]
    score += evaluate_line(anti, player, opponent) - evaluate_line(anti, opponent, player)
    return score

def game_board(n):
    board = []
    for i in range(n):
        row = []
        for j in range(n):
            row.append(0)
        board.append(row)
    return board


def winner(board):
    n = len(board)
    for row in board:
        if row[0] != 0 and all(cell == row[0] for cell in row):
            return row[0]
    for col in range(n):
        first = board[0][col]
        if first != 0 and all(board[row][col] == first for row in range(n)):
            return first
    first = board[0][0]
    if first != 0 and all(board[i][i] == first for i in range(n)):
        return first
    first = board[0][n - 1]
    if first != 0 and all(board[i][n - 1 - i] == first for i in range(n)):
        return first
    if any(cell == 0 for row in board for cell in row):
        return -1
    return 0


def simple_player_move(board, me, opponent):
    n = len(board)
    for row in range(n):
        for col in range(n):
            if board[row][col] == 0:
                board[row][col] = me
                if winner(board) == me:
                    return
                board[row][col] = 0
    for row in range(n):
        for col in range(n):
            if board[row][col] == 0:
                board[row][col] = opponent
                blocks = winner(board) == opponent
                board[row][col] = 0
                if blocks:
                    board[row][col] = me
                    return
    center = n // 2
    if n % 2 == 1 and board[center][center] == 0:
        board[center][center] = me
        return
    for row in range(n):
        for col in range(n):
            if board[row][col] == 0:
                board[row][col] = me
                return
                
def minimax(board, depth, max_depth, turn, me, opponent, alpha, beta):
    status = winner(board)
    if status == me:
        return 1000 - depth
    if status == opponent:
        return -1000 + depth
    if status == 0:
        return 0
    if depth == max_depth:
        return heuristic_score(board, me, opponent)

    n = len(board)
    next_turn = opponent if turn == me else me
    is_max = (turn == me)
    best = float('-inf') if is_max else float('inf')
    for row in range(n):
        for col in range(n):
            if board[row][col] == 0:
                board[row][col] = turn
                score = minimax(board, depth + 1, max_depth, next_turn, me, opponent, alpha, beta)
                board[row][col] = 0
                if is_max:
                    best = max(best, score); alpha = max(alpha, best)
                else:
                    best = min(best, score); beta = min(beta, best)
                if beta <= alpha:
                    return best
    return best

def smart_player_move(board, me, opponent, max_depth):
    n = len(board)
    best_score, best_move = float('-inf'), None
    for row in range(n):
        for col in range(n):
            if board[row][col] == 0:
                board[row][col] = me
                score = minimax(board, 1, max_depth, opponent, me, opponent, float('-inf'), float('inf'))
                board[row][col] = 0
                if score > best_score:
                    best_score, best_move = score, (row, col)
    row, col = best_move
    board[row][col] = me

def play_game(n, max_depth, smart_starts):
    board = game_board(n)
    SMART, SIMPLE = 1, 2
    turn = SMART if smart_starts else SIMPLE
    while True:
        if turn == SMART:
            smart_player_move(board, SMART, SIMPLE, max_depth)
        else:
            simple_player_move(board, SIMPLE, SMART)
        status = winner(board)
        if status != -1:
            return status
        turn = SIMPLE if turn == SMART else SMART

def run_experiment(n, max_depth, num_games=10):
    results = {"smart_wins": 0, "simple_wins": 0, "draws": 0}
    for i in range(num_games):
        status = play_game(n, max_depth, smart_starts=(i % 2 == 0)) 
        key = {1: "smart_wins", 2: "simple_wins", 0: "draws"}[status]
        results[key] += 1
    return results

for n, depth in [(3, 9), (4, 4), (5, 3)]:
    print(n, run_experiment(n, depth))


````

````python
def watch_game(n=3, max_depth=9, smart_starts=True):
    board = game_board(n)
    SMART, SIMPLE = 1, 2
    turn = SMART if smart_starts else SIMPLE

    print("Starting board:")
    draw_game_board(board)

    move_num = 1
    while True:
        if turn == SMART:
            smart_player_move(board, SMART, SIMPLE, max_depth)
            print(f"\nMove {move_num}: Smart player (X) moves")
        else:
            simple_player_move(board, SIMPLE, SMART)
            print(f"\nMove {move_num}: Simple player (O) moves")

        draw_game_board(board)

        status = winner(board)
        if status == SMART:
            print("Smart player (X) wins!")
            return
        elif status == SIMPLE:
            print("Simple player (O) wins!")
            return
        elif status == 0:
            print("It's a draw!")
            return

        turn = SIMPLE if turn == SMART else SMART
        move_num += 1


watch_game(n=3, max_depth=9, smart_starts=True)


````

- Writing small, focused functions
- Using conditionals and loops to analyze program state
- Thinking carefully about assumptions and edge cases
- Using LLMs **responsibly** as learning tools rather than answer generators

The goal is not just to make the program work, but to understand *why* it works.
That understanding is what allows you to use tools — including AI — effectively.
