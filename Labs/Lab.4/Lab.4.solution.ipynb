1. Write a "counter" class that can be incremented up to a specified maximum value, will print an error if an attempt is made to increment beyond that value, and allows reseting the counter. 

```python
class Counter:
    def __init__(self, max_value):
        self.max_value = max_value
        self.count = 0
    def increment(self):
        if self.count < self.max_value:
            self.count += 1
        else:
            print(f"Error: counter cannot exceed maximum value of {self.max_value}")
    def reset(self):
        self.count = 0
```

```python
import math

class Shape:
    def __init__(self, x, y):
        self.__x = x
        self.__y = y
    def get_x(self):
        return self.__x
    def get_y(self):
        return self.__y
    def area(self):
        raise NotImplementedError
    def perimeter(self):
        raise NotImplementedError
    def perimeter_points(self, num_points=16):
        raise NotImplementedError
    def is_inside(self, px, py):
        raise NotImplementedError
    def _points_on_polygon(self, vertices, num_points):
        num_points = min(num_points, 16)
        n = len(vertices)
        side_lengths = [
            math.hypot(vertices[(i + 1) % n][0] - vertices[i][0],
                       vertices[(i + 1) % n][1] - vertices[i][1])
            for i in range(n)
        ]
        total = sum(side_lengths)

        points = []
        for i in range(num_points):
            target = total * i / num_points
            cumulative = 0
            for side_index in range(n):
                side_len = side_lengths[side_index]
                if target <= cumulative + side_len or side_index == n - 1:
                    start = vertices[side_index]
                    end = vertices[(side_index + 1) % n]
                    fraction = (target - cumulative) / side_len if side_len > 0 else 0
                    px = start[0] + fraction * (end[0] - start[0])
                    py = start[1] + fraction * (end[1] - start[1])
                    points.append((px, py))
                    break
                cumulative += side_len
        return points


class Rectangle(Shape):
    def __init__(self, length, width, x, y):
        super().__init__(x, y)
        self.__length = length
        self.__width = width
    def get_length(self):
        return self.__length
    def get_width(self):
        return self.__width
    def area(self):
        return self.__length * self.__width
    def perimeter(self):
        return 2 * (self.__length + self.__width)
    def perimeter_points(self, num_points=16):
        x, y = self.get_x(), self.get_y()
        vertices = [
            (x, y),
            (x + self.__length, y),
            (x + self.__length, y + self.__width),
            (x, y + self.__width),
        ]
        return self._points_on_polygon(vertices, num_points)
    def is_inside(self, px, py):
        x, y = self.get_x(), self.get_y()
        return (x <= px <= x + self.__length) and (y <= py <= y + self.__width)


class Circle(Shape):
    def __init__(self, radius, x, y):
        super().__init__(x, y)
        self.__radius = radius
    def get_radius(self):
        return self.__radius
    def area(self):
        return math.pi * self.__radius ** 2
    def perimeter(self):
        return 2 * math.pi * self.__radius
    def perimeter_points(self, num_points=16):
        num_points = min(num_points, 16)
        x, y = self.get_x(), self.get_y()
        r = self.__radius
        points = []
        for i in range(num_points):
            theta = 2 * math.pi * i / num_points
            points.append((x + r * math.cos(theta), y + r * math.sin(theta)))
        return points
    def is_inside(self, px, py):
        x, y = self.get_x(), self.get_y()
        distance = math.hypot(px - x, py - y)
        return distance <= self.__radius


class Triangle(Shape):
    def __init__(self, side_a, side_b, side_c, x, y):
        super().__init__(x, y)
        self.__side_a = side_a
        self.__side_b = side_b
        self.__side_c = side_c
    def get_side_a(self):
        return self.__side_a
    def get_side_b(self):
        return self.__side_b
    def get_side_c(self):
        return self.__side_c
    def perimeter(self):
        return self.__side_a + self.__side_b + self.__side_c
    def area(self):
        a, b, c = self.__side_a, self.__side_b, self.__side_c
        s = self.perimeter() / 2
        return math.sqrt(s * (s - a) * (s - b) * (s - c))
    def _get_vertices(self):
        x, y = self.get_x(), self.get_y()
        a, b, c = self.__side_a, self.__side_b, self.__side_c
        v1 = (x, y)
        v2 = (x + c, y)
        cos_angle = max(-1, min(1, (b**2 + c**2 - a**2) / (2 * b * c)))
        angle = math.acos(cos_angle)
        v3 = (x + b * math.cos(angle), y + b * math.sin(angle))
        return [v1, v2, v3]
    def perimeter_points(self, num_points=16):
        return self._points_on_polygon(self._get_vertices(), num_points)
    def is_inside(self, px, py):
        v1, v2, v3 = self._get_vertices()
        def sign(p1, p2, p3):
            return (p1[0] - p3[0]) * (p2[1] - p3[1]) - (p2[0] - p3[0]) * (p1[1] - p3[1])

        d1 = sign((px, py), v1, v2)
        d2 = sign((px, py), v2, v3)
        d3 = sign((px, py), v3, v1)

        has_neg = (d1 < 0) or (d2 < 0) or (d3 < 0)
        has_pos = (d1 > 0) or (d2 > 0) or (d3 > 0)
        return not (has_neg and has_pos)
```

```python
import math

class Shape:
    def __init__(self, x, y):
        self.__x = x
        self.__y = y
    def get_x(self):
        return self.__x
    def get_y(self):
        return self.__y
    def area(self):
        raise NotImplementedError
    def perimeter(self):
        raise NotImplementedError
    def perimeter_points(self, num_points=16):
        raise NotImplementedError
    def is_inside(self, px, py):
        raise NotImplementedError
    def overlaps(self, other):
        for (px, py) in self.perimeter_points():
            if other.is_inside(px, py):
                return True
        for (px, py) in other.perimeter_points():
            if self.is_inside(px, py):
                return True
        return False
    def _points_on_polygon(self, vertices, num_points):
        num_points = min(num_points, 16)
        n = len(vertices)
        side_lengths = [
            math.hypot(vertices[(i + 1) % n][0] - vertices[i][0],
                       vertices[(i + 1) % n][1] - vertices[i][1])
            for i in range(n)
        ]
        total = sum(side_lengths)
        points = []
        for i in range(num_points):
            target = total * i / num_points
            cumulative = 0
            for side_index in range(n):
                side_len = side_lengths[side_index]
                if target <= cumulative + side_len or side_index == n - 1:
                    start = vertices[side_index]
                    end = vertices[(side_index + 1) % n]
                    fraction = (target - cumulative) / side_len if side_len > 0 else 0
                    px = start[0] + fraction * (end[0] - start[0])
                    py = start[1] + fraction * (end[1] - start[1])
                    points.append((px, py))
                    break
                cumulative += side_len
        return points


class Rectangle(Shape):
    def __init__(self, length, width, x, y):
        super().__init__(x, y)
        self.__length = length
        self.__width = width
    def get_length(self):
        return self.__length
    def get_width(self):
        return self.__width
    def area(self):
        return self.__length * self.__width
    def perimeter(self):
        return 2 * (self.__length + self.__width)
    def perimeter_points(self, num_points=16):
        x, y = self.get_x(), self.get_y()
        vertices = [
            (x, y),
            (x + self.__length, y),
            (x + self.__length, y + self.__width),
            (x, y + self.__width),
        ]
        return self._points_on_polygon(vertices, num_points)

    def is_inside(self, px, py):
        x, y = self.get_x(), self.get_y()
        return (x <= px <= x + self.__length) and (y <= py <= y + self.__width)


class Circle(Shape):
    def __init__(self, radius, x, y):
        super().__init__(x, y)
        self.__radius = radius
    def get_radius(self):
        return self.__radius
    def area(self):
        return math.pi * self.__radius ** 2
    def perimeter(self):
        return 2 * math.pi * self.__radius
    def perimeter_points(self, num_points=16):
        num_points = min(num_points, 16)
        x, y = self.get_x(), self.get_y()
        r = self.__radius
        points = []
        for i in range(num_points):
            theta = 2 * math.pi * i / num_points
            points.append((x + r * math.cos(theta), y + r * math.sin(theta)))
        return points
    def is_inside(self, px, py):
        x, y = self.get_x(), self.get_y()
        distance = math.hypot(px - x, py - y)
        return distance <= self.__radius


class Triangle(Shape):
    def __init__(self, side_a, side_b, side_c, x, y):
        super().__init__(x, y)
        self.__side_a = side_a
        self.__side_b = side_b
        self.__side_c = side_c
    def get_side_a(self):
        return self.__side_a
    def get_side_b(self):
        return self.__side_b
    def get_side_c(self):
        return self.__side_c
    def perimeter(self):
        return self.__side_a + self.__side_b + self.__side_c
    def area(self):
        a, b, c = self.__side_a, self.__side_b, self.__side_c
        s = self.perimeter() / 2
        return math.sqrt(s * (s - a) * (s - b) * (s - c))
    def _get_vertices(self):
        x, y = self.get_x(), self.get_y()
        a, b, c = self.__side_a, self.__side_b, self.__side_c
        v1 = (x, y)
        v2 = (x + c, y)
        cos_angle = max(-1, min(1, (b**2 + c**2 - a**2) / (2 * b * c)))
        angle = math.acos(cos_angle)
        v3 = (x + b * math.cos(angle), y + b * math.sin(angle))
        return [v1, v2, v3]
    def perimeter_points(self, num_points=16):
        return self._points_on_polygon(self._get_vertices(), num_points)
    def is_inside(self, px, py):
        v1, v2, v3 = self._get_vertices()
        def sign(p1, p2, p3):
            return (p1[0] - p3[0]) * (p2[1] - p3[1]) - (p2[0] - p3[0]) * (p1[1] - p3[1])
        d1 = sign((px, py), v1, v2)
        d2 = sign((px, py), v2, v3)
        d3 = sign((px, py), v3, v1)
        has_neg = (d1 < 0) or (d2 < 0) or (d3 < 0)
        has_pos = (d1 > 0) or (d2 > 0) or (d3 > 0)
        return not (has_neg and has_pos)
```

```python
class RasterDrawing:
    def __init__(self, width, height):
        self.__canvas = Canvas(width, height)
        self.__shapes = []  
    def add_shape(self, shape, char='*'):
        self.__shapes.append((shape, char))
    def remove_shape(self, shape):
        self.__shapes = [(s, c) for (s, c) in self.__shapes if s is not shape]
    def get_shapes(self):
        return [s for (s, c) in self.__shapes]
    def paint(self):

        self.__canvas.clear_canvas()
        for shape, char in self.__shapes:
            shape.paint(self.__canvas, char=char)
    def display(self):
        self.__canvas.display()
```

```python
drawing = RasterDrawing(50, 32)
body = Rectangle(20, 12, 10, 5)
roof = Triangle(12.806, 12.806, 20, 10, 17)
house = CompoundShape([body, roof])
sun = Circle(4, 22, 24)
drawing.add_shape(house, char='#')
drawing.add_shape(sun, char='o')
print("Original drawing (house + sun):")
drawing.paint()
drawing.display()
drawing.remove_shape(sun)
trunk = Rectangle(2, 6, 34, 5)
leaves = Circle(5, 35, 15)
tree = CompoundShape([trunk, leaves])
drawing.add_shape(tree, char='%')

print()
print("Modified drawing (sun removed, tree added):")
drawing.paint()
drawing.display()
```

```python
import math


class Canvas:
    def __init__(self, width, height):
        self.width = width
        self.height = height
        self.data = [[' '] * width for i in range(height)]

    def set_pixel(self, row, col, char='*'):
        self.data[row][col] = char

    def get_pixel(self, row, col):
        return self.data[row][col]

    def clear_canvas(self):
        self.data = [[' '] * self.width for i in range(self.height)]

    def v_line(self, x, y, w, **kargs):
        for i in range(x, x + w):
            self.set_pixel(i, y, **kargs)

    def h_line(self, x, y, h, **kargs):
        for i in range(y, y + h):
            self.set_pixel(x, i, **kargs)

    def line(self, x1, y1, x2, y2, **kargs):
        slope = (y2 - y1) / (x2 - x1)
        for y in range(y1, y2):
            x = int(slope * y)
            self.set_pixel(x, y, **kargs)

    def display(self):
        print("\n".join(["".join(row) for row in self.data]))


class Shape:
    def __init__(self, x, y):
        self.__x = x
        self.__y = y

    def get_x(self):
        return self.__x

    def get_y(self):
        return self.__y

    def area(self):
        raise NotImplementedError

    def perimeter(self):
        raise NotImplementedError

    def perimeter_points(self, num_points=16):
        raise NotImplementedError

    def is_inside(self, px, py):
        raise NotImplementedError

    def overlaps(self, other):
        for (px, py) in self.perimeter_points():
            if other.is_inside(px, py):
                return True
        for (px, py) in other.perimeter_points():
            if self.is_inside(px, py):
                return True
        return False

    def paint(self, canvas, char='*'):
        points = self.perimeter_points()
        n = len(points)
        for i in range(n):
            x1, y1 = points[i]
            x2, y2 = points[(i + 1) % n]
            self._plot_segment(canvas, x1, y1, x2, y2, char)

    def _to_grid(self, canvas, x, y):
        row = canvas.height - 1 - round(y)
        col = round(x)
        return row, col

    def _plot_segment(self, canvas, x1, y1, x2, y2, char):
        r1, c1 = self._to_grid(canvas, x1, y1)
        r2, c2 = self._to_grid(canvas, x2, y2)
        steps = max(abs(r2 - r1), abs(c2 - c1), 1)
        for step in range(steps + 1):
            r = round(r1 + (r2 - r1) * step / steps)
            c = round(c1 + (c2 - c1) * step / steps)
            if 0 <= r < canvas.height and 0 <= c < canvas.width:
                canvas.set_pixel(r, c, char)

    def _points_on_polygon(self, vertices, num_points):
        num_points = min(num_points, 16)
        n = len(vertices)
        side_lengths = [
            math.hypot(vertices[(i + 1) % n][0] - vertices[i][0],
                       vertices[(i + 1) % n][1] - vertices[i][1])
            for i in range(n)
        ]
        total = sum(side_lengths)
        points = []
        for i in range(num_points):
            target = total * i / num_points
            cumulative = 0
            for side_index in range(n):
                side_len = side_lengths[side_index]
                if target <= cumulative + side_len or side_index == n - 1:
                    start = vertices[side_index]
                    end = vertices[(side_index + 1) % n]
                    fraction = (target - cumulative) / side_len if side_len > 0 else 0
                    px = start[0] + fraction * (end[0] - start[0])
                    py = start[1] + fraction * (end[1] - start[1])
                    points.append((px, py))
                    break
                cumulative += side_len
        return points


class Rectangle(Shape):
    def __init__(self, length, width, x, y):
        super().__init__(x, y)
        self.__length = length
        self.__width = width

    def get_length(self):
        return self.__length

    def get_width(self):
        return self.__width

    def area(self):
        return self.__length * self.__width

    def perimeter(self):
        return 2 * (self.__length + self.__width)

    def perimeter_points(self, num_points=16):
        x, y = self.get_x(), self.get_y()
        vertices = [
            (x, y),
            (x + self.__length, y),
            (x + self.__length, y + self.__width),
            (x, y + self.__width),
        ]
        return self._points_on_polygon(vertices, num_points)

    def is_inside(self, px, py):
        x, y = self.get_x(), self.get_y()
        return (x <= px <= x + self.__length) and (y <= py <= y + self.__width)

    def __repr__(self):
        return f"Rectangle({self.__length!r}, {self.__width!r}, {self.get_x()!r}, {self.get_y()!r})"


class Circle(Shape):
    def __init__(self, radius, x, y):
        super().__init__(x, y)
        self.__radius = radius

    def get_radius(self):
        return self.__radius

    def area(self):
        return math.pi * self.__radius ** 2

    def perimeter(self):
        return 2 * math.pi * self.__radius

    def perimeter_points(self, num_points=16):
        num_points = min(num_points, 16)
        x, y = self.get_x(), self.get_y()
        r = self.__radius
        points = []
        for i in range(num_points):
            theta = 2 * math.pi * i / num_points
            points.append((x + r * math.cos(theta), y + r * math.sin(theta)))
        return points

    def is_inside(self, px, py):
        x, y = self.get_x(), self.get_y()
        distance = math.hypot(px - x, py - y)
        return distance <= self.__radius

    def __repr__(self):
        return f"Circle({self.__radius!r}, {self.get_x()!r}, {self.get_y()!r})"


class Triangle(Shape):
    def __init__(self, side_a, side_b, side_c, x, y):
        super().__init__(x, y)
        self.__side_a = side_a
        self.__side_b = side_b
        self.__side_c = side_c

    def get_side_a(self):
        return self.__side_a

    def get_side_b(self):
        return self.__side_b

    def get_side_c(self):
        return self.__side_c

    def perimeter(self):
        return self.__side_a + self.__side_b + self.__side_c

    def area(self):
        a, b, c = self.__side_a, self.__side_b, self.__side_c
        s = self.perimeter() / 2
        return math.sqrt(s * (s - a) * (s - b) * (s - c))
        
    def _get_vertices(self):
        x, y = self.get_x(), self.get_y()
        a, b, c = self.__side_a, self.__side_b, self.__side_c
        v1 = (x, y)
        v2 = (x + c, y)
        cos_angle = max(-1, min(1, (b**2 + c**2 - a**2) / (2 * b * c)))
        angle = math.acos(cos_angle)
        v3 = (x + b * math.cos(angle), y + b * math.sin(angle))
        return [v1, v2, v3]
        
    def perimeter_points(self, num_points=16):
        return self._points_on_polygon(self._get_vertices(), num_points)
        
    def is_inside(self, px, py):
        v1, v2, v3 = self._get_vertices()

        def sign(p1, p2, p3):
            return (p1[0] - p3[0]) * (p2[1] - p3[1]) - (p2[0] - p3[0]) * (p1[1] - p3[1])
        d1 = sign((px, py), v1, v2)
        d2 = sign((px, py), v2, v3)
        d3 = sign((px, py), v3, v1)
        has_neg = (d1 < 0) or (d2 < 0) or (d3 < 0)
        has_pos = (d1 > 0) or (d2 > 0) or (d3 > 0)
        return not (has_neg and has_pos)
    def __repr__(self):
        return (f"Triangle({self.__side_a!r}, {self.__side_b!r}, {self.__side_c!r}, "
                f"{self.get_x()!r}, {self.get_y()!r})")


class CompoundShape(Shape):
    def __init__(self, shapes):
        if not shapes:
            raise ValueError
        super().__init__(shapes[0].get_x(), shapes[0].get_y())
        self.__shapes = list(shapes)
        
    def get_shapes(self):
        return list(self.__shapes)
        
    def area(self):
        return sum(shape.area() for shape in self.__shapes)
        
    def perimeter(self):
        return sum(shape.perimeter() for shape in self.__shapes)
        
    def perimeter_points(self, num_points=16):
        num_points = min(num_points, 16)
        per_shape = max(1, num_points // len(self.__shapes))
        points = []
        for shape in self.__shapes:
            points.extend(shape.perimeter_points(per_shape))
        return points
        
    def is_inside(self, px, py):
        return any(shape.is_inside(px, py) for shape in self.__shapes)
        
    def overlaps(self, other):
        others = other.get_shapes() if isinstance(other, CompoundShape) else [other]
        return any(shape.overlaps(o) for shape in self.__shapes for o in others)
        
    def paint(self, canvas, char='*'):
        for shape in self.__shapes:
            shape.paint(canvas, char=char)
            
    def __repr__(self):
        return f"CompoundShape({self.__shapes!r})"


class RasterDrawing:
    def __init__(self, width, height, shapes=None):
        self.__width = width
        self.__height = height
        self.__canvas = Canvas(width, height)
        self.__shapes = list(shapes) if shapes else []  # (shape, char) pairs
    def add_shape(self, shape, char='*'):
        self.__shapes.append((shape, char))
    def remove_shape(self, shape):
        self.__shapes = [(s, c) for (s, c) in self.__shapes if s is not shape]
    def get_shapes(self):
        return [s for (s, c) in self.__shapes]
    def paint(self):
        self.__canvas.clear_canvas()
        for shape, char in self.__shapes:
            shape.paint(self.__canvas, char=char)
    def display(self):
        self.__canvas.display()
    def __repr__(self):
        return f"RasterDrawing({self.__width!r}, {self.__height!r}, {self.__shapes!r})"
    def save(self, filename):
        with open(filename, "w") as f:
            f.write(repr(self))


def raster_drawing_loader(filename):
    with open(filename, "r") as f:
        return eval(f.read())

```
