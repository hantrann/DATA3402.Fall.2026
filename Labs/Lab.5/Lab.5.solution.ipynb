1. Create a `matrix` class with the following properties:
    * It can be initialized in 2 ways:
        1. with arguments `n` and `m`, the size of the matrix. A newly instanciated matrix will contain all zeros.
        2. with a list of lists of values. Note that since we are using lists of lists to implement matrices, it is possible that not all rows have the same number of columns. Test explicitly that the matrix is properly specified.
    * Matrix instances `M` can be indexed with `M[i][j]` and `M[i,j]`.
    * Matrix assignment works in 2 ways:
        1. If `M_1` and `M_2` are `matrix` instances `M_1=M_2` sets the values of `M_1` to those of `M_2`, if they are the same size. Error otherwise.
        2. In example above `M_2` can be a list of lists of correct size.


```python
class matrix:
    def __init__(self, n=None, m=None):
        if isinstance(n, int) and isinstance(m, int):
            self.data = []
            for i in range(n):
                row = []
                for j in range(m):
                    row.append(0)
                self.data.append(row)


        elif isinstance(n, list) and m is None:
            if len(n) == 0:
                raise ValueError
            columns = len(n[0])
            for row in n:
                if not isinstance(row, list):
                    raise TypeError
                if len(row) != columns:
                    raise ValueError
            self.data = []
            for row in n:
                self.data.append(row.copy())
        else:
            raise ValueError
            
    def __getitem__(self, index):
        if isinstance(index, tuple):
            i, j = index
            return self.data[i][j]
        return self.data[index]

    def __setitem__(self, index, value):
        if isinstance(index, tuple):
            i, j = index
            self.data[i][j] = value
        else:
            self.data[index] = value

    def assign(self, other):
        if isinstance(other, matrix):
            source = other.data
        elif isinstance(other, list):
            source = other

        else:
            raise TypeError
        if len(source) != len(self.data):
            raise ValueError
        for i in range(len(self.data)):
            if len(source[i]) != len(self.data[i]):
                raise ValueError
        for i in range(len(self.data)):
            for j in range(len(self.data[i])):
                self.data[i][j] = source[i][j]
```

2. Add the following methods:
    * `shape()`: returns a tuple `(n,m)` of the shape of the matrix.
    * `transpose()`: returns a new matrix instance which is the transpose of the matrix.
    * `row(n)` and `column(n)`: that return the nth row or column of the matrix M as a new appropriately shaped matrix object.
    * `to_list()`: which returns the matrix as a list of lists.
    *  `block(n_0,n_1,m_0,m_1)` that returns a smaller matrix located at the n_0 to n_1 columns and m_0 to m_1 rows. 
    * Modify `__getitem__` implemented above to support slicing.
        

```python
class matrix:
    def __init__(self, n=None, m=None):
        if isinstance(n, int) and isinstance(m, int):
            self.data = []
            for i in range(n):
                row = []
                for j in range(m):
                    row.append(0)
                self.data.append(row)

        elif isinstance(n, list) and m is None:
            if len(n) == 0:
                raise ValueError
            columns = len(n[0])
            for row in n:
                if not isinstance(row, list):
                    raise TypeError
                if len(row) != columns:
                    raise ValueError
            self.data = []
            for row in n:
                self.data.append(row.copy())
        else:
            raise ValueError
            
    def __getitem__(self, index):
        if isinstance(index, tuple):
            row_index, column_index = index
            if isinstance(row_index, slice):
                rows = self.data[row_index]
            else:
                rows = [self.data[row_index]]
            result = []
            for row in rows:
                if isinstance(column_index, slice):
                    result.append(row[column_index])
                else:
                    result.append([row[column_index]])

            if not isinstance(row_index, slice) and not isinstance(column_index, slice):
                return result[0][0]
            return matrix(result)
        return self.data[index]

    def __setitem__(self, index, value):
        if isinstance(index, tuple):
            i, j = index
            self.data[i][j] = value
        else:
            self.data[index] = value

    def shape(self):
        return (len(self.data), len(self.data[0]))

    def transpose(self):
        n, m = self.shape()
        result = []
        for j in range(m):
            row = []
            for i in range(n):
                row.append(self.data[i][j])
            result.append(row)
        return matrix(result)

    def row(self, n):
        return matrix([self.data[n].copy()])

    def column(self, n):
        result = []
        for row in self.data:
            result.append([row[n]])
        return matrix(result)

    def to_list(self):
        result = []
        for row in self.data:
            result.append(row.copy())
        return result

    def block(self, n_0, n_1, m_0, m_1):
        result = []
        for i in range(m_0, m_1):
            row = []
            for j in range(n_0, n_1):
                row.append(self.data[i][j])
            result.append(row)
        return matrix(result)
    
```

3. Write functions that create special matrices (note these are standalone functions, not member functions of your `matrix` class):
    * `constant(n,m,c)`: returns a `n` by `m` matrix filled with floats of value `c`.
    * `zeros(n,m)` and `ones(n,m)`: return `n` by `m` matrices filled with floats of value `0` and `1`, respectively.
    * `eye(n)`: returns the n by n identity matrix.

```python
def constant(n, m, c):
    data = []
    for i in range(n):
        row = []
        for j in range(m):
            row.append(float(c))
        data.append(row)
    return matrix(data)

def zeros(n, m):
    return constant(n, m, 0)

def ones(n, m):
    return constant(n, m, 1)

def eye(n):
    data = []
    for i in range(n):
        row = []
        for j in range(n):
            if i == j:
                row.append(1.0)
            else:
                row.append(0.0)
        data.append(row)
    return matrix(data)
```

```python

class matrix:
    def __init__(self, n=None, m=None):
        if isinstance(n, int) and isinstance(m, int):
            self.data = []
            for i in range(n):
                row = []
                for j in range(m):
                    row.append(0)
                self.data.append(row)

        elif isinstance(n, list) and m is None:
            if len(n) == 0:
                raise ValueError
            columns = len(n[0])
            for row in n:
                if not isinstance(row, list):
                    raise TypeError
                if len(row) != columns:
                    raise ValueError
            self.data = []
            for row in n:
                self.data.append(row.copy())
        else:
            raise ValueError
            
    def __getitem__(self, index):
        if isinstance(index, tuple):
            row_index, column_index = index
            if isinstance(row_index, slice):
                rows = self.data[row_index]
            else:
                rows = [self.data[row_index]]
            result = []
            for row in rows:
                if isinstance(column_index, slice):
                    result.append(row[column_index])
                else:
                    result.append([row[column_index]])

            if not isinstance(row_index, slice) and not isinstance(column_index, slice):
                return result[0][0]
            return matrix(result)
        return self.data[index]

    def __setitem__(self, index, value):
        if isinstance(index, tuple):
            i, j = index
            self.data[i][j] = value
        else:
            self.data[index] = value

    def shape(self):
        return (len(self.data), len(self.data[0]))

    def transpose(self):
        n, m = self.shape()
        result = []
        for j in range(m):
            row = []
            for i in range(n):
                row.append(self.data[i][j])
            result.append(row)
        return matrix(result)

    def row(self, n):
        return matrix([self.data[n].copy()])

    def column(self, n):
        result = []
        for row in self.data:
            result.append([row[n]])
        return matrix(result)

    def to_list(self):
        result = []
        for row in self.data:
            result.append(row.copy())
        return result

    def block(self, n_0, n_1, m_0, m_1):
        result = []
        for i in range(m_0, m_1):
            row = []
            for j in range(n_0, n_1):
                row.append(self.data[i][j])
            result.append(row)
        return matrix(result)





    def scalarmul(self, c):
        n, m = self.shape()
        result = []
        for i in range(n):
            row = []
            for j in range(m):
                row.append(c * self.data[i][j])
            result.append(row)
        return matrix(result)

    def add(self, N):
        if not isinstance(N, matrix):
            raise TypeError("N must be a matrix")
        if self.shape() != N.shape():
            raise ValueError("Matrices must have the same size")
        n, m = self.shape()
        result = []
        for i in range(n):
            row = []
            for j in range(m):
                row.append(self.data[i][j] + N.data[i][j])
            result.append(row)
        return matrix(result)

    def sub(self, N):
        if not isinstance(N, matrix):
            raise TypeError("N must be a matrix")
        if self.shape() != N.shape():
            raise ValueError("Matrices must have the same size")
        n, m = self.shape()
        result = []
        for i in range(n):
            row = []
            for j in range(m):
                row.append(self.data[i][j] - N.data[i][j])
            result.append(row)
        return matrix(result)

    def mat_mult(self, N):
        if not isinstance(N, matrix):
            raise TypeError("N must be a matrix")
        n1, m1 = self.shape()
        n2, m2 = N.shape()
        if m1 != n2:
            raise ValueError(
                "Number of columns of first matrix must equal "
                "number of rows of second matrix"
            )

        result = []
        for i in range(n1):
            row = []
            for j in range(m2):
                value = 0
                for k in range(m1):
                    value += self.data[i][k] * N.data[k][j]
                row.append(value)
            result.append(row)
        return matrix(result)

    def element_mult(self, N):
        if not isinstance(N, matrix):
            raise TypeError("N must be a matrix")
        if self.shape() != N.shape():
            raise ValueError("Matrices must have the same size")
        n, m = self.shape()
        result = []
        for i in range(n):
            row = []
            for j in range(m):
                row.append(self.data[i][j] * N.data[i][j])
            result.append(row)
        return matrix(result)

    def equals(self, N):
        if not isinstance(N, matrix):
            return False
        if self.shape() != N.shape():
            return False
        n, m = self.shape()
        for i in range(n):
            for j in range(m):
                if self.data[i][j] != N.data[i][j]:
                    return False

        return True
```

```python
class matrix:
    def __init__(self, n=None, m=None):
        if isinstance(n, int) and isinstance(m, int):
            self.data = []
            for i in range(n):
                row = []
                for j in range(m):
                    row.append(0)
                self.data.append(row)

        elif isinstance(n, list) and m is None:
            if len(n) == 0:
                raise ValueError
            columns = len(n[0])
            for row in n:
                if not isinstance(row, list):
                    raise TypeError
                if len(row) != columns:
                    raise ValueError
            self.data = []
            for row in n:
                self.data.append(row.copy())
        else:
            raise ValueError
            
    def __getitem__(self, index):
        if isinstance(index, tuple):
            row_index, column_index = index
            if isinstance(row_index, slice):
                rows = self.data[row_index]
            else:
                rows = [self.data[row_index]]
            result = []
            for row in rows:
                if isinstance(column_index, slice):
                    result.append(row[column_index])
                else:
                    result.append([row[column_index]])

            if not isinstance(row_index, slice) and not isinstance(column_index, slice):
                return result[0][0]
            return matrix(result)
        return self.data[index]

    def __setitem__(self, index, value):
        if isinstance(index, tuple):
            i, j = index
            self.data[i][j] = value
        else:
            self.data[index] = value

    def shape(self):
        return (len(self.data), len(self.data[0]))

    def transpose(self):
        n, m = self.shape()
        result = []
        for j in range(m):
            row = []
            for i in range(n):
                row.append(self.data[i][j])
            result.append(row)
        return matrix(result)

    def row(self, n):
        return matrix([self.data[n].copy()])

    def column(self, n):
        result = []
        for row in self.data:
            result.append([row[n]])
        return matrix(result)

    def to_list(self):
        result = []
        for row in self.data:
            result.append(row.copy())
        return result

    def block(self, n_0, n_1, m_0, m_1):
        result = []
        for i in range(m_0, m_1):
            row = []
            for j in range(n_0, n_1):
                row.append(self.data[i][j])
            result.append(row)
        return matrix(result)





    def scalarmul(self, c):
        n, m = self.shape()
        result = []
        for i in range(n):
            row = []
            for j in range(m):
                row.append(c * self.data[i][j])
            result.append(row)
        return matrix(result)

    def add(self, N):
        if not isinstance(N, matrix):
            raise TypeError("N must be a matrix")
        if self.shape() != N.shape():
            raise ValueError("Matrices must have the same size")
        n, m = self.shape()
        result = []
        for i in range(n):
            row = []
            for j in range(m):
                row.append(self.data[i][j] + N.data[i][j])
            result.append(row)
        return matrix(result)

    def sub(self, N):
        if not isinstance(N, matrix):
            raise TypeError("N must be a matrix")
        if self.shape() != N.shape():
            raise ValueError("Matrices must have the same size")
        n, m = self.shape()
        result = []
        for i in range(n):
            row = []
            for j in range(m):
                row.append(self.data[i][j] - N.data[i][j])
            result.append(row)
        return matrix(result)

    def mat_mult(self, N):
        if not isinstance(N, matrix):
            raise TypeError("N must be a matrix")
        n1, m1 = self.shape()
        n2, m2 = N.shape()
        if m1 != n2:
            raise ValueError(
                "Number of columns of first matrix must equal "
                "number of rows of second matrix"
            )

        result = []
        for i in range(n1):
            row = []
            for j in range(m2):
                value = 0
                for k in range(m1):
                    value += self.data[i][k] * N.data[k][j]
                row.append(value)
            result.append(row)
        return matrix(result)

    def element_mult(self, N):
        if not isinstance(N, matrix):
            raise TypeError("N must be a matrix")
        if self.shape() != N.shape():
            raise ValueError("Matrices must have the same size")
        n, m = self.shape()
        result = []
        for i in range(n):
            row = []
            for j in range(m):
                row.append(self.data[i][j] * N.data[i][j])
            result.append(row)
        return matrix(result)

    def equals(self, N):
        if not isinstance(N, matrix):
            return False
        if self.shape() != N.shape():
            return False
        n, m = self.shape()
        for i in range(n):
            for j in range(m):
                if self.data[i][j] != N.data[i][j]:
                    return False

        return True




    def __mul__(self, other):
        if isinstance(other, matrix):
            return self.mat_mult(other)
        elif isinstance(other, (int, float)):
            return self.scalarmul(other)
        else:
            raise TypeError

    def __rmul__(self, other):
        if isinstance(other, (int, float)):
            return self.scalarmul(other)
        else:
            raise TypeError

    def __add__(self, other):
        return self.add(other)

    def __sub__(self, other):
        return self.sub(other)

    def __eq__(self, other):
        return self.equals(other)


    
def constant(n, m, c):
    data = []
    for i in range(n):
        row = []
        for j in range(m):
            row.append(float(c))
        data.append(row)
    return matrix(data)

def zeros(n, m):
    return constant(n, m, 0)

def ones(n, m):
    return constant(n, m, 1)

def eye(n):
    data = []
    for i in range(n):
        row = []
        for j in range(n):
            if i == j:
                row.append(1.0)
            else:
                row.append(0.0)
        data.append(row)
    return matrix(data)
```

```python
A = matrix([
    [1, 2],
    [3, 4]
])

B = matrix([
    [2, 0],
    [1, 2]
])

C = matrix([
    [1, 1],
    [0, 1]
])

I = eye(2)
```

```python
left = (A * B) * C
right = A * (B * C)
print("\n1. Associative Property")
print("(AB)C =", left.to_list())
print("A(BC)  =", right.to_list())
print("(AB)C == A(BC):", left == right)

left = A * (B + C)
right = (A * B) + (A * C)
print("\n2. Distributive Property")
print("A(B+C) =", left.to_list())
print("AB+AC  =", right.to_list())
print("A(B+C) == AB+AC:", left == right)


AB = A * B
BA = B * A
print("\n3. Non-Commutative Property")
print("AB =", AB.to_list())
print("BA =", BA.to_list())
print("AB != BA:", not (AB == BA))

AI = A * I
print("\n4. Identity Property")
print("A  =", A.to_list())
print("AI =", AI.to_list())
print("AI == A:", AI == A)
```
